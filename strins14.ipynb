```python
# String Operators

a = "Hello"
b = "Python"

# Concatenation operator (+)
print("Concatenation:", a + " " + b)

# Repetition operator (*)
print("Repetition:", a * 3)

# Membership operator (in)
print("'H' in a:", "H" in a)

# Membership operator (not in)
print("'z' not in a:", "z" not in a)

# Comparison operators
print("Equal:", a == b)
print("Not Equal:", a != b)
print("Greater Than:", a > b)
print("Less Than:", a < b)
```

```python
# String Operations

text = "Python Programming"

# Convert to uppercase
print("Uppercase:", text.upper())

# Convert to lowercase
print("Lowercase:", text.lower())

# Capitalize the first character
print("Capitalize:", text.capitalize())

# Convert each word to title case
print("Title Case:", text.title())

# Replace a word
print("Replace:", text.replace("Python", "Java"))

# Split a string into a list
print("Split:", text.split())

# Check whether the string starts with a word
print("Starts With:", text.startswith("Python"))

# Check whether the string ends with a word
print("Ends With:", text.endswith("Programming"))

# Remove leading and trailing spaces
message = "  Hello Python  "
print("Strip:", message.strip())

# Find the position of a substring
print("Find:", text.find("Program"))
```

```python
# String Built-in Functions and Methods

text = "Python123"

# Find the length
print("Length:", len(text))

# Find the maximum character
print("Maximum:", max(text))

# Find the minimum character
print("Minimum:", min(text))

# Count occurrences of a character
print("Count of 'o':", text.count("o"))

# Find the index of a character
print("Index of 't':", text.index("t"))

# Check whether all characters are letters
print("Is Alphabetic:", text.isalpha())

# Check whether all characters are digits
print("Is Numeric:", text.isdigit())

# Check whether the string contains only letters and numbers
print("Is Alphanumeric:", text.isalnum())

# Check whether all letters are uppercase
print("Is Uppercase:", text.isupper())

# Check whether all letters are lowercase
print("Is Lowercase:", text.islower())
```
